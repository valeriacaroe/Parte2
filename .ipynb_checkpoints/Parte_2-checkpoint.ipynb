### Exercise 5

<div class="alert-success">
    
1. Get a polygons map of the lowest administrative unit possible.

2. Get a table of variables for those units. At least 3 numerical variables.

3. Preprocess both tables and get them ready for merging.

4. Do the merging, making the changes needed so that you keep the most rows.

```python
import geopandas as gpd
depas=gpd.read_file('https://github.com/valeriacaroe/Parte2/raw/refs/heads/main/departamento/departamentoPolygon.shp')
pobreza=gpd.read_file('https://github.com/valeriacaroe/Parte2/raw/refs/heads/main/Pobreza%20Cr%C3%B3nica%20x%20Departamento/Pobreza%20Cr%C3%B3nica%20X%20Departamento.shp')
```

```python
pobreza.columns
```

```python
pobreza=pobreza.drop(columns=['LINK_DEPTO','PROVINCIA_','DEPARTAM_1','DEPARTAM_2','DEPARTAM_3','DEPARTAM_4' ])
```

```python
pobreza
```

```python
pobreza.info()
```

```python
clean_column_names = lambda df: df.columns.astype(str).str.strip().str.replace(pat=r'\s', repl='_', regex=True)

pobreza.columns = clean_column_names(pobreza)
```

```python
pobreza=pobreza[['DEPTO_FIRS', 'N_HOGARES', 'POBLACION','P_PC_HOG', 'geometry']]
```

```python
depas
```

```python
depas=depas[['fna']]
```

```python
depas
```

```python
onlyPobreza=set(pobreza.DEPTO_FIRS)-set(depas.fna)
onlyPobreza
```

```python
onlyDepas=set(depas.fna)-set(pobreza.DEPTO_FIRS)
onlyDepas
```

```python
!pip install thefuzz -q
from thefuzz import process as fz
```

```python
[(item, fz.extractOne(item,onlyDepas)) for item in sorted(onlyPobreza)]
```

```python
# Se ve que la mayoría de discordancias entre nombres se da cuando la concordancia es menor a 86
[(item, fz.extractOne(item,onlyDepas)) for item in sorted(onlyPobreza)
 if fz.extractOne(item, onlyDepas)[1]>86]
```

```python
changes1={item:fz.extractOne(item, onlyDepas)[0]
                 for item in sorted(onlyPobreza)
                 if fz.extractOne(item, onlyDepas)[1] >86}
#dict of matches
changes1
```

```python
changes1_limpio = {}
for key, value in changes1.items():
    if value not in changes1_limpio.values():
        changes1_limpio[key] = value

changes1 = changes1_limpio
```

```python
changes1
```

```python
pobreza.replace(to_replace={'DEPTO_FIRS':changes1},inplace=True)
```

Merge

```python
merge=depas.merge(pobreza,left_on='fna',right_on='DEPTO_FIRS')
merge.drop(columns=['DEPTO_FIRS'],inplace=True)
merge
```

```python
merge = merge.set_geometry('geometry')
merge.plot()
```

2. Make a scatter plot for each variable.

```python
%pip install esda -q
from esda.moran import Moran
```

```python
x = merge['N_HOGARES'].values
```

```python
w = KNN.from_dataframe(merge, k=8)
w.transform = 'r'
```

```python
moran = Moran(x, w)
```

```python
print(f"Moran's I: {moran.I:.4f}")
print(f"p-value: {moran.p_sim:.4f}")
```

```python
%pip install splot -q
from splot.esda import moran_scatterplot
```

```python
fig, ax = moran_scatterplot(moran, aspect_equal=True)
ax.set_xlabel('N_Hogares')
ax.set_ylabel('SpatialLag_Hogares_std');
```

```python
moranPoblacion = Moran(merge['POBLACION'], w)
moranPoblacion.I,moranPoblacion.p_sim
fig, ax = moran_scatterplot(moranPoblacion, aspect_equal=True)
ax.set_xlabel('Población')
ax.set_ylabel('SpatialLag_Poblacion_std');
```

```python
moran3 = Moran(merge['P_PC_HOG'], w)
moran3.I,moran3.p_sim
fig, ax = moran_scatterplot(moran3, aspect_equal=True)
ax.set_xlabel('P_PC_HOG')
ax.set_ylabel('SpatialLag_PCHog_std');
```

### Exercise 8

<div class="alert-success">
    
1. Compute the Local Moran for the variables in your data that have significant spatial correlation.
    
2. Create a new column for each of those variables, with a label ('0 no_sig', '1 hotSpot', '2 coldOutlier', '3 coldSpot', '4 hotOutlier').

3. Prepare a map for each of the variables analyzed, showing the spots and outliers.

```python
merge.head()
```

```python
# A LISA for each district using IDH2019
from esda.moran import Moran_Local
lisaIDH = Moran_Local(y=merge['N_HOGARES'], w=w,seed=2022)
```

```python
fig, ax = moran_scatterplot(lisaIDH,p=0.05)
ax.set_xlabel('N_HOGARES_std')
ax.set_ylabel('SpatialLag_N_HOGARES_std');
```

```python
lisaIDH.q, lisaIDH.p_sim
```

```python
import pandas as pd
pd.Series(lisaIDH.q).value_counts()
```

```python
merge['N_HOG_quadrant']=[l if p <0.05 else 0 for l,p in zip(lisaIDH.q,lisaIDH.p_sim)  ]
merge['N_HOG_quadrant'].value_counts()
```

```python
labels = [ '0 no_sig', '1 hotSpot', '2 coldOutlier', '3 coldSpot', '4 hotOutlier']

merge['HOGARES_quadrant_names']=[labels[i] for i in merge['N_HOG_quadrant']]

merge['HOGARES_quadrant_names'].value_counts()
```

```python
# custom colors
from matplotlib import colors
import matplotlib.pyplot as plt

myColMap = colors.ListedColormap([ 'white', 'pink', 'cyan', 'azure','red'])

# Set up figure and ax
f, ax = plt.subplots(1, figsize=(12,12))
# Plot unique values choropleth including
# a legend and with no boundary lines

plt.title('Spots and Outliers')

merge.plot(column='HOGARES_quadrant_names',
                categorical=True,
                cmap=myColMap,
                linewidth=0.1,
                edgecolor='k',
                legend=True,
                legend_kwds={'loc': 'center left',
                             'bbox_to_anchor': (0.7, 0.6)},
                ax=ax)
# Remove axis
ax.set_axis_off()
# Display the map
plt.show()
```

Exercise 9
Use your three variables to carry out the cluster/regional analysis.

```python
merge.head()
```

```python
!pip install mapclassify
```

```python
import mapclassify
import numpy as np
np.random.seed(12345)

theVar=merge.N_HOGARES
K=5
#
ei5 = mapclassify.EqualInterval(theVar, k=K)
#
msd = mapclassify.StdMean(theVar)
#
q5=mapclassify.Quantiles(theVar,k=K)
#
mb5 = mapclassify.MaximumBreaks(theVar, k=K)
#
ht = mapclassify.HeadTailBreaks(theVar)
#
fj5 = mapclassify.FisherJenks(theVar, k=K)
#
jc5 = mapclassify.JenksCaspall(theVar, k=K)
#
mp5 = mapclassify.MaxP(theVar, k=K)
```

```python
import seaborn
class5 = q5, ei5,msd, ht, mb5, fj5, jc5, mp5
# Collect ADCM for each classifier
fits = np.array([ c.adcm for c in class5])
# Convert ADCM scores to a DataFrame
adcms = pd.DataFrame(fits)
# Add classifier names
adcms['classifier'] = [c.name for c in class5]
# Add column names to the ADCM
adcms.columns = ['ADCM', 'Classifier']
adcms.sort_values(by='ADCM', ascending=False).plot(kind='bar', x='Classifier', y='ADCM')
```

```python
merge['N_HOGARES']=fj5.yb
```

```python
theVar=merge.POBLACION
K=5
#
ei51 = mapclassify.EqualInterval(theVar, k=K)
#
msd1 = mapclassify.StdMean(theVar)
#
q51=mapclassify.Quantiles(theVar,k=K)
#
mb51 = mapclassify.MaximumBreaks(theVar, k=K)
#
ht1 = mapclassify.HeadTailBreaks(theVar)
#
fj51 = mapclassify.FisherJenks(theVar, k=K)
#
jc51 = mapclassify.JenksCaspall(theVar, k=K)
#
mp51 = mapclassify.MaxP(theVar, k=K)
```

```python
class51 = q51, ei51,msd1, ht1, mb51, fj51, jc51, mp51
# Collect ADCM for each classifier
fits = np.array([ c.adcm for c in class51])
# Convert ADCM scores to a DataFrame
adcms = pd.DataFrame(fits)
# Add classifier names
adcms['classifier'] = [c.name for c in class51]
# Add column names to the ADCM
adcms.columns = ['ADCM', 'Classifier']
adcms.sort_values(by='ADCM', ascending=False).plot(kind='bar', x='Classifier', y='ADCM')
```

```python
selected_variables = ['N_HOGARES',
                     'POBLACION',
                     'P_PC_HOG']
```

```python
import seaborn as sns
sns.boxplot(merge[selected_variables])
```

```python
merge[selected_variables].corr()
```

```python
import seaborn as sns
sns.pairplot(
    merge[selected_variables], kind="reg", diag_kind="kde"
)
```

```python
from sklearn.preprocessing import StandardScaler


scaler = StandardScaler()
normalized_data = scaler.fit_transform(merge[selected_variables])
sns.displot(pd.melt(pd.DataFrame(normalized_data,columns=selected_variables)),
            x="value", hue="variable",kind="kde",
            log_scale=(False,False))
```

```python
selected_variables_new_std=[s+'_std' for s in selected_variables]

# add colunms
merge[selected_variables_new_std]=normalized_data
```

 Conventional Clustering


```python
from scipy.cluster import hierarchy as hc


Z = hc.linkage(merge[selected_variables_new_std], 'ward')
# calculate full dendrogram
plt.figure(figsize=(25, 10))
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('cases')
plt.ylabel('distance')
hc.dendrogram(
    Z,
    leaf_rotation=90.,  # rotates the x axis labels
    leaf_font_size=1,  # font size for the x axis labels
)
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering as agnes

import numpy as np
np.random.seed(12345)# Set seed for reproducibility

model = agnes(linkage="ward", n_clusters=6).fit(merge[selected_variables_new_std])

# Assign labels to main data table
merge["hc_ag6"] = model.labels_
```

```python
merge["hc_ag6"].value_counts()
```

```python
merge.groupby("hc_ag6")[selected_variables_new_std].mean()
```

PRIMER INTENTO DE FORMAR LOS CLUSTER

```python
# Set up figure and ax
f, ax = plt.subplots(1, figsize=(9, 9))
# Plot unique values choropleth including
# a legend and with no boundary lines
merge.plot(
    column="hc_ag6", categorical=True, legend=True, linewidth=0, ax=ax
)
# Remove axis
ax.set_axis_off()
# Display the map
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering as agnes
from libpysal.weights import Queen

w_queen_merge = Queen.from_dataframe(merge)

model_queen = agnes(linkage="ward",
                    n_clusters=6,
                    connectivity=w_queen_merge.sparse).fit(merge[selected_variables_new_std])
#
merge["hc_ag6_wQueen"] = model_queen.labels_
```

```python
# Set up figure and ax
f, ax = plt.subplots(1, figsize=(9, 9))
# Plot unique values choropleth including
# a legend and with no boundary lines
merge.plot(
    column="hc_ag6_wQueen", categorical=True, legend=True, linewidth=0, ax=ax
)
# Remove axis
ax.set_axis_off()
# Display the map
plt.show()
```
